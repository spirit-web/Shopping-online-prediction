# Data Cleaning – Online Shopping Prediction

I den här notebooken kontrollerar jag datakvaliteten i de svenska datasetfilerna innan analys och supervised machine learning.  
Syftet är att undersöka om datan innehåller saknade värden, dubletter, fel datatyper eller oväntade värden.

```python
import pandas as pd
import numpy as np
```

```python
train_df = pd.read_csv("../../data/training_sample_svenska.csv")
test_df = pd.read_csv("../../data/testing_sample_svenska.csv")

print("Training shape:", train_df.shape)
print("Testing shape:", test_df.shape)
```

```python
train_df.head()
```

```python
test_df.head()
```

Datasetet innehåller beteendedata från en webbshop. Varje rad representerar en användare eller session, och kolumnerna visar olika handlingar, till exempel om besökaren klickade på varukorgen, loggade in, gick till kassan eller genomförde en beställning.

```python
train_df.info()
```

```python
test_df.info()
```

De flesta variabler är numeriska och verkar vara binära, alltså 0 eller 1. Kolumnen `anvandar_id` är en textkolumn och används inte som prediktor i machine learning-modellen.

```python
missing_train = train_df.isnull().sum()
missing_test = test_df.isnull().sum()

print("Missing values in training data:")
print(missing_train[missing_train > 0])

print("\nMissing values in testing data:")
print(missing_test[missing_test > 0])
```

Det finns inga saknade värden i datasetet. Därför behövs ingen imputation eller borttagning av rader på grund av missing values.

```python
print("Duplicate rows in training data:", train_df.duplicated().sum())
print("Duplicate rows in testing data:", test_df.duplicated().sum())
```

```python
print("Duplicate user IDs in training data:", train_df["anvandar_id"].duplicated().sum())
print("Duplicate user IDs in testing data:", test_df["anvandar_id"].duplicated().sum())
```

Jag kontrollerade både hela dublettrader och dubletter i `anvandar_id`. Om det inte finns dubletter behöver inga rader tas bort.

```python
train_df["genomfort_bestallning"].value_counts()
```

```python
train_df["genomfort_bestallning"].value_counts(normalize=True) * 100
```

Målvariabeln `genomfort_bestallning` innehåller två klasser: 0 och 1. Datasetet är tydligt obalanserat eftersom majoriteten av besökarna inte genomför en beställning. Detta är viktigt att komma ihåg vid modellutvärdering, eftersom accuracy kan bli missvisande.

```python
feature_cols = train_df.drop(columns=["anvandar_id", "genomfort_bestallning"]).columns

for col in feature_cols:
    unique_values = sorted(train_df[col].unique())
    print(col, unique_values)
```

```python
for col in feature_cols:
    unique_values = sorted(test_df[col].unique())
    print(col, unique_values)
```

Alla feature-variabler verkar vara binära, alltså 0 eller 1. Det betyder att datan redan är kodad på ett sätt som passar bra för supervised machine learning.

```python
train_clean = train_df.copy()
test_clean = test_df.copy()

train_clean.head()
```

```python
train_clean.to_csv("../../data/training_sample_svenska_cleaned.csv", index=False)
test_clean.to_csv("../../data/testing_sample_svenska_cleaned.csv", index=False)
```
